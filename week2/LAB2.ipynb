```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier
```

```python
from sklearn.metrics import(
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay

)
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/data.csv")
```

```python
df.head()
```

```python

df['Attendance'] = [68, 75, 77, 89, 80, 72, 91, 65, 76]
```

```python
print(df.info)
```

```python
df.isnull()
```

```python
df.shape
```

```python
df['Attendance'] = df['Attendance'].fillna(df['Attendance'].mode()[0])
```

```python
df.isnull().sum()
```

```python
X=df[['Age','StudyHours','Attendance','Marks']]
y=df['Result']
```

```python
encoder=LabelEncoder()
y=encoder.fit_transform(y)
print(y)
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,y,
    test_size=0.2,
    random_state=32,
    stratify=y
)
```

```python
print(len(X_train))
print(len(y_train))

print(len(X_test))
print(len(y_test))

```

```python
#Create the classifier

model=DecisionTreeClassifier(random_state=42)
```

```python
model.fit(X_train,y_train)
```

```python
df.head()
```

```python
y_pred=model.predict(X_test)

print("Actual",y_test)
print("Predicted",y_pred)
```

```python
accuracy=accuracy_score(y_test,y_pred)
print(accuracy*100)
```

```python
precision=precision_score(y_test,y_pred)
print(precision)
```

```python
df.head()
```

```python
print(recall_score(y_test,y_pred,zero_division=0))
```

```python
print(f1_score(y_test,y_pred,zero_division=0))
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=encoder.transform(encoder.classes_)
)

print("Confusion Matrix")
print(cm)
```

```python
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=encoder.classes_
)

disp.plot()

plt.title("Confusion Matrix")
plt.show()
```
