```python
import pandas as pd

# Read dataset
data = pd.read_csv("/content/sample_data/car_evaluation.csv")

# Column names
data.columns = [
    'buying',
    'maint',
    'doors',
    'persons',
    'lug_boot',
    'safety',
    'class'
]

features = [
    'buying',
    'maint',
    'doors',
    'persons',
    'lug_boot',
    'safety'
]

target = 'class'

rules = []

# Go through every class
for cls in data[target].unique():

    print("\nFinding rules for:", cls)

    # Positive examples of current class
    remaining = data[data[target] == cls].copy()

    while len(remaining) > 0:

        best_feature = None
        best_value = None

        # IMPORTANT FIX
        best_score = float('-inf')

        # Try every feature
        for feature in features:

            # Try every value
            for value in data[feature].unique():

                # Positive examples still remaining
                positive = len(
                    remaining[remaining[feature] == value]
                )

                if positive == 0:
                    continue

                # All examples satisfying this condition
                covered = data[data[feature] == value]

                # Examples from other classes
                negative = len(
                    covered[covered[target] != cls]
                )

                # Score
                score = positive - negative

                # Select best condition
                if score > best_score:
                    best_score = score
                    best_feature = feature
                    best_value = value

        # No rule found
        if best_feature is None:
            break

        # Create rule
        rule = (
            f"IF {best_feature} = {best_value} "
            f"THEN class = {cls}"
        )

        rules.append(rule)

        print(rule)

        # Remove covered positive examples
        remaining = remaining[
            remaining[best_feature] != best_value
        ]


print("\n========== FINAL RULES ==========")

for i, rule in enumerate(rules, 1):
    print(f"Rule {i}: {rule}")
```
